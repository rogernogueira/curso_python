## 1. Os verbos do texto

`.strip()` apara os espaços das pontas; `.title()` põe a primeira letra de cada palavra em maiúscula **e as demais em minúscula**;
`.lower()` põe tudo em minúscula; `.replace(velho, novo)` troca trechos. Encadeados, formam uma linha de produção.

````python
nome = "  ana ribeiro nogueira  "
print(nome.strip())
print(nome.strip().title())
print("prazo vencido".replace("vencido", "cumprido"))
````

O texto é **imutável**: o verbo devolve um texto novo e o original fica igual. Para ficar com a versão limpa, guarde o resultado.

````python
nome = "  ana ribeiro nogueira  "
nome.strip()
print(repr(nome))
````

````python
nome = "  ana ribeiro nogueira  "
nome_limpo = nome.strip().title()
print(repr(nome_limpo))
````

> **No Colab:** se `nome.strip()` for a última linha da célula, a célula mostra o texto limpo, mas é só o texto novo que o verbo devolveu: `nome` continua sujo, como prova o `print(repr(nome))`.

O status sujo da pauta, `"Concluso "`, limpo antes da decisão:

````python
def texto_decisao(status):
    if status == "concluso":
        return "Autos conclusos. Decido."
    elif status == "prazo vencido":
        return "Certifique-se o decurso do prazo e voltem conclusos."
    elif status == "aguardando":
        return "Aguarde-se o cumprimento. Após, conclusos."
    elif status == "arquivado":
        return "Arquivem-se os autos."
    else:
        return "Situação não prevista: verificar."

status = "Concluso ".strip().lower()
print(repr(status))
print(texto_decisao(status))
````

## 2. A pinça: índices e fatias

O texto é uma fileira de caracteres numerados a partir do zero. Na fatia `texto[início:fim]`, o início entra e o fim
não entra, a mesma regra do `range`.

````python
numero_processo = "0004512-33.2026.8.27.0001"
print(numero_processo[0])
print(numero_processo[0:7])
print(numero_processo[11:15])
````

Quando há separador, o `.split()` é mais robusto: ele fatia o texto numa lista, e `[1]` é o segundo pedaço, porque a lista também conta a partir do zero.

````python
numero_processo = "0004512-33.2026.8.27.0001"
print(numero_processo.split("."))
print(numero_processo.split(".")[1])
````

## 3. A f-string: o modelo com lacunas

Com `f` antes das aspas, cada `{...}` é preenchido com o valor. `:.2f` dá as duas casas do dinheiro.
Dentro de uma f-string de aspas duplas, as chaves do dicionário usam aspas simples.

> Aspas iguais dentro e fora funcionam a partir do Python 3.12; o caderno, escrito para versões anteriores, diz que dão erro.
> Alternar as aspas funciona nos dois casos. Para saber a versão do seu Colab: `!python --version`.

````python
processo = {
    "numero_processo": "0004512-33.2026.8.27.0001",
    "nome_parte": "  ana ribeiro nogueira ",
    "vara": "1ª Vara Cível",
    "valor_causa": 15400.0,
}

nome = processo["nome_parte"].strip().title()
valor = f"R$ {processo['valor_causa']:.2f}"
print(f"Processo {processo['numero_processo']} — {nome} — {valor}")
````

## 4. A `montar_despacho` definitiva

As aspas triplas deixam o texto ocupar várias linhas, exatamente como vai sair, espaços incluídos: por isso as linhas
do modelo começam na margem, mesmo dentro da função. O valor sai com ponto (`103420.00`) porque o `:.2f` segue o padrão
americano. A ficha da seção 4.4 do caderno não tem `"status"` e daria `KeyError`; aqui a ficha tem o campo.

````python
def texto_decisao(status):
    if status == "concluso":
        return "Autos conclusos. Decido."
    elif status == "prazo vencido":
        return "Certifique-se o decurso do prazo e voltem conclusos."
    elif status == "aguardando":
        return "Aguarde-se o cumprimento. Após, conclusos."
    elif status == "arquivado":
        return "Arquivem-se os autos."
    else:
        return "Situação não prevista: verificar."

def montar_despacho(processo):
    nome = processo["nome_parte"].strip().title()
    status = processo["status"].strip().lower()
    decisao = texto_decisao(status)
    return f"""Processo nº {processo['numero_processo']}
Requerente: {nome}
Vara: {processo['vara']}
Valor da causa: R$ {processo['valor_causa']:.2f}

DESPACHO

{decisao}"""

processo = {
    "numero_processo": "0007397-50.2025.8.27.0001",
    "nome_parte": "  maria de fátima gomes  ",
    "vara": "1ª Vara Cível",
    "status": "aguardando",
    "valor_causa": 103420.0,
}

print(montar_despacho(processo))
````
