## 1. Dicionário: a ficha com campos nomeados

Um **dicionário** guarda pares de chave e valor. A chave vai entre aspas porque é um texto (o nome do campo); `processo`,
sem aspas, é a variável que guarda a ficha inteira. Esta ficha troca o valor da causa da lição 1 pelo `status`, o campo
que a decisão da lição 2 usa.

````python
processo = {
    "numero_processo": "0004512-33.2026.8.27.0001",
    "nome_parte": "Ana Ribeiro Nogueira",
    "vara": "1ª Vara Cível",
    "status": "concluso",
}

print(processo["nome_parte"])
print(processo.get("vara"))
for chave, valor in processo.items():
    print(chave, "=", valor)
````

## 2. Função: a máquina entra-processa-sai

`def` define a função; o **parâmetro** é o dado que entra; o `return` entrega o resultado e encerra a função.
Definir não é executar: a célula abaixo não mostra nada.

````python
def texto_decisao(status):
    if status == "concluso":
        return "Autos conclusos. Decido."
    elif status == "prazo vencido":
        return "Certifique-se o decurso do prazo e voltem conclusos."
    elif status == "aguardando":
        return "Aguarde-se o cumprimento. Após, conclusos."
    elif status == "arquivado":
        return "Arquivem-se os autos."
    else:
        return "Situação não prevista: verificar."
````

A função trabalha quando é chamada pelo nome, com parênteses:

````python
print(texto_decisao("prazo vencido"))
print(texto_decisao("suspenso"))
````

> **Quem estuda pelo caderno:** lá, o `else` arquiva e a frase do decurso é a antiga. Aqui seguimos a versão corrigida na lição 2:
> `elif` próprio para "arquivado", `else` que avisa e a redação "Certifique-se o decurso do prazo e voltem conclusos."

## 4. A função que monta o despacho

Uma função chama a outra. O texto ainda é montado com `+` e `\n` (quebra de linha): funciona, mas é um rascunho honesto;
a lição 4 o deixa bonito. Os parênteses em volta da montagem só deixam a conta continuar na linha de baixo.

````python
def texto_decisao(status):
    if status == "concluso":
        return "Autos conclusos. Decido."
    elif status == "prazo vencido":
        return "Certifique-se o decurso do prazo e voltem conclusos."
    elif status == "aguardando":
        return "Aguarde-se o cumprimento. Após, conclusos."
    elif status == "arquivado":
        return "Arquivem-se os autos."
    else:
        return "Situação não prevista: verificar."

def montar_despacho(processo):
    decisao = texto_decisao(processo["status"])
    texto = ("Processo " + processo["numero_processo"] + "\n"
             + "Requerente: " + processo["nome_parte"] + "\n\n"
             + "DESPACHO\n" + decisao)
    return texto

processo = {
    "numero_processo": "0004512-33.2026.8.27.0001",
    "nome_parte": "Ana Ribeiro Nogueira",
    "vara": "1ª Vara Cível",
    "status": "concluso",
}

print(montar_despacho(processo))
````
