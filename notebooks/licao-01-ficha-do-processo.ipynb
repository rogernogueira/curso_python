## 2. Variável: a caixinha com nome

Pense na capa do processo no eproc: campos com nome (Número do Processo, Partes, Órgão Julgador, Valor da Causa),
cada um com um conteúdo. Em Python, cada campo vira uma **variável**: uma caixinha com nome que guarda um valor.
O sinal `=` não pergunta nada: ele **guarda** o valor da direita na caixinha da esquerda.

O `print` com vírgulas mostra cada item separado por um espaço. **Antes de executar, pense: o que vai aparecer?**

````python
numero_processo = "0004512-33.2026.8.27.0001"
nome_parte = "Ana Ribeiro Nogueira"
vara = "1ª Vara Cível"
valor_causa = 15400.00
print(nome_parte, "—", numero_processo)
````

Os nomes das caixinhas seguem três cuidados:

- tudo em minúsculas, com traço baixo no lugar do espaço (**snake_case**): `nome_parte`, `valor_causa`;
- espaço no nome é proibido: `nome parte` dá erro;
- maiúsculas contam: `Vara` e `vara` são caixinhas diferentes.

## 3. Tipo: a moldura das aspas

Com aspas é **texto** (`str`); sem aspas é **número**. Número com casa decimal, escrito com ponto, é `float`;
número sem casa decimal, como 3, é `int` (ele aparece no exercício). A pergunta que decide não é "tem dígitos?",
e sim "**vou fazer conta com isso?**". Ninguém soma números de processo: texto. O valor da causa entra nas custas: decimal.

As caixinhas criadas lá em cima continuam na memória; é por isso que se executa de cima para baixo. Se aparecer
**NameError** (erro de nome: a caixinha não existe), execute as células anteriores ou use *Executar tudo*.

Para perguntar o tipo, use `print(type(...))`. **Antes de executar, pense: o que vai aparecer?**

````python
print(type(nome_parte))
print(type(valor_causa))
````

## 4. A conta das custas

`custas = valor_causa * 0.01` faz 15.400,00 × 0,01 = 154,00 (o `*` multiplica). A linha que começa com `#` é um
comentário: o Python a ignora, e ela serve para explicar o porquê. **Antes de executar, pense: o que vai aparecer?**

````python
# Custas iniciais: 1% sobre o valor da causa (tabela da comarca)
custas = valor_causa * 0.01
print("Custas iniciais: R$", custas)
````

Sai `154.0`, com uma casa: é assim que o Python mostra o número. As duas casas do dinheiro (154,00) chegam na lição 4.

## 5. Um erro, lido com calma

Quando o Python para por erro, ele mostra uma mensagem em vermelho chamada **traceback** (rastreamento). Lê-se de
baixo para cima: a **última linha** diz o tipo do erro e o que aconteceu.

Na célula abaixo, apague o `#` e o espaço do começo da linha, execute, leia a última linha e ponha o `#` de volta.

````python
# idade_processo = "3" + 2
````

A última linha é:

```
TypeError: can only concatenate str (not "int") to str
```

**TypeError** (erro de tipo) quer dizer operação entre tipos que não combinam: o texto `"3"` não se soma ao número `2`.
No Colab o cabeçalho do traceback tem outro formato; a última linha é a mesma.

O conserto: `int()` transforma o texto `"3"` no número 3, e aí a soma funciona.

````python
idade_processo = int("3") + 2
print(idade_processo)
````
