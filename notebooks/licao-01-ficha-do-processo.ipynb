## 3. Tipo: a moldura das aspas

Com aspas é **texto** (`str`); sem aspas é **número**: inteiro (`int`) ou decimal (`float`, escrito com ponto).
A pergunta que decide não é "tem dígitos?", e sim "**vou fazer conta com isso?**". Ninguém soma números de processo: texto.
O valor da causa entra nas custas: decimal. Para perguntar o tipo, use `print(type(...))`.

````python
numero_processo = "0004512-33.2026.8.27.0001"
nome_parte = "Ana Ribeiro Nogueira"
vara = "1ª Vara Cível"
valor_causa = 15400.00
print(type(nome_parte))
print(type(valor_causa))
print(type(numero_processo))
print(type(12))
````

Quando o dado chega como texto, a função com o nome do tipo de destino converte:

````python
valor_texto = "1500"
valor_numero = float(valor_texto)
print(valor_numero, type(valor_numero))
````

A linha que começa com `#` é um comentário: o Python a ignora, e ela serve para explicar o porquê.

## 5. Traceback: a certidão lida de baixo para cima

Quando o Python para por erro, ele mostra o **traceback** (rastreamento). Lê-se de baixo para cima: a última linha diz o
**tipo do erro**; as de cima, **onde**. **TypeError** (erro de tipo) quer dizer operação entre tipos que não combinam.

Somar o texto `"3"` com o número `2` não funciona. Descomente a célula abaixo (apague os `# `) e execute. A última linha do erro, executada de verdade, é:

````python
# Erro de propósito: descomente para ver o erro, depois comente de novo.
# idade_processo = "3" + 2
````

````python
idade_processo = int("3") + 2
print(idade_processo)
````

## 6. O mundo real: o valor que chega como "15.400,00"

Copiado da planilha de controle da secretaria, o valor vem no formato brasileiro e como texto.
`float("15.400,00")` recusa com **ValueError** (erro de valor): o tipo serve, mas o formato não.
Tirar as aspas é pior: o Python **não reclama** e guarda outra coisa.

````python
valor_causa = 15.400,00
print(valor_causa, type(valor_causa))
````

Veio `(15.4, 0)`, um par de números (uma **tupla**, que volta na lição 3), não o valor. O jeito certo: tratar o texto e só então converter. O `.replace()` é tema da lição 4.

````python
texto = "15.400,00"
valor_causa = float(texto.replace(".", "").replace(",", "."))
print(valor_causa, type(valor_causa))
print(valor_causa * 0.01)
````
