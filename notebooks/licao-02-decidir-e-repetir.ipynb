## 1. Comparar: `==` pergunta, `=` guarda

Uma comparação responde `True` (verdadeiro) ou `False` (falso): esse é o tipo **booleano** (`bool`).
`and` exige as duas condições; `or` aceita uma; `not` inverte. Na ficha da Ana:

````python
valor_causa = 15400.00
vara = "1ª Vara Cível"

print(valor_causa > 10000)
print(not (valor_causa > 10000))
print(valor_causa > 10000 and vara == "1ª Vara Cível")
````

## 2. Os incisos: `if`, `elif`, `else`

O Python testa as hipóteses de cima para baixo e executa **só a primeira** verdadeira; o `else` é a cláusula residual.
Os dois-pontos no fim e o recuo das linhas de dentro são obrigatórios.

````python
status = "prazo vencido"

if status == "concluso":
    frase = "Autos conclusos. Decido."
elif status == "prazo vencido":
    frase = "Certifico o decurso do prazo. Tornem os autos conclusos."
elif status == "aguardando":
    frase = "Aguarde-se o cumprimento. Após, conclusos."
else:
    frase = "Arquivem-se os autos."

print(frase)
````

## 3. A prateleira e a fila que anda sozinha

Uma **lista** guarda vários valores em ordem, entre colchetes. O `for` percorre a lista e repete o bloco recuado para cada item.

````python
fila = ["concluso", "prazo vencido", "aguardando", "arquivado"]
print(fila)
print(len(fila))
````

````python
fila = ["concluso", "prazo vencido", "aguardando", "arquivado"]

for status in fila:
    if status == "concluso":
        frase = "Autos conclusos. Decido."
    elif status == "prazo vencido":
        frase = "Certifico o decurso do prazo. Tornem os autos conclusos."
    elif status == "aguardando":
        frase = "Aguarde-se o cumprimento. Após, conclusos."
    else:
        frase = "Arquivem-se os autos."
    print(status, "->", frase)
````

Para numerar, o `enumerate` entrega o número e o item; com `start=1`, começa do 1:

````python
fila = ["concluso", "prazo vencido", "aguardando", "arquivado"]
for i, status in enumerate(fila, start=1):
    print(i, status)
````

## 4. O `range`: o início entra, o fim não entra

Como um marca-página: o fim marca onde parar, sem ler aquela página. `range(1, 3)` produz 1 e 2.

> **Atenção:** não compare o `range` à contagem de prazos. O art. 224 do CPC, salvo disposição em contrário, faz o contrário:
> exclui o dia do começo e inclui o do vencimento. Por isso o curso usa o marca-página. (O caderno usa essa analogia; aqui ela foi trocada.)

````python
for n in range(1, 3):
    print(n)
````

## 6. Mundo real: o `else` que arquiva o que não conhece

No código acima, o `"suspenso"` e o `"Concluso "` caem no `else` e seriam arquivados. A versão defensável tem uma hipótese
própria para `"arquivado"` e um `else` que **avisa** em vez de arquivar:

````python
fila = ["concluso", "prazo vencido", "aguardando", "arquivado", "suspenso"]

for i, status in enumerate(fila, start=1):
    if status == "concluso":
        frase = "Autos conclusos. Decido."
    elif status == "prazo vencido":
        frase = "Certifico o decurso do prazo. Tornem os autos conclusos."
    elif status == "aguardando":
        frase = "Aguarde-se o cumprimento. Após, conclusos."
    elif status == "arquivado":
        frase = "Arquivem-se os autos."
    else:
        frase = "Situação não prevista: verificar."
    print(i, status, "->", frase)
````

### Sua vez: a triagem do protocolo

Leia o código e preveja: quantas linhas da saída terão o destino **Distribuição**?

````python
fila = ["inicial", "recurso", "intermediária", "inicial", "contestação", "recurso"]

for tipo in fila:
    if tipo == "inicial":
        destino = "Distribuição"
    elif tipo == "recurso":
        destino = "Setor de recursos"
    elif tipo == "intermediária":
        destino = "Juntada aos autos"
    else:
        destino = "Triagem manual"
    print(tipo, "->", destino)
````
